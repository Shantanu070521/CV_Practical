```python
from sympy import symbols, diff, solve, Matrix
```

```python
x, y = symbols('x y')
f = y**2 - x**2 - 4*x - 6*y + 13
print(f'The function is: f(x,y) = {f}')
```

```python
fx = diff(f, x)
fy = diff(f, y)
print(f'First derivative (fx):{fx}')
print(f'First derivative (fy):{fy}')
```

```python
solution=solve((fx, fy), (x, y))
print(f'The solution is: {solution}')
```

```python
stationary_points = solve([fx, fy], (x, y))

print(f"Stationary points: {stationary_points}")
```

```python
fxx = diff(fx, x)
fyy = diff(fy, y)
fxy = diff(fx, y)
print(f"d2f/dx2 = {fxx}")
print(f"d2f/dy2 = {fyy}")
print(f"d2f/dxdy = {fxy}")
```

```python
H=Matrix([[fxx,fxy],[fxy,fyy]])
print("\nHessian Matrix H:")
display(H)
```

```python
det_H = H.det()
print(f"Determinant of Hessian (det(H)): {det_H}")

if det_H > 0:
    if fxx > 0:
        print("The stationary point is a local minimum.")
    elif fxx < 0:
        print("The stationary point is a local maximum.")
else:
    print("The Hessian test is inconclusive (det(H) <= 0).")
```
